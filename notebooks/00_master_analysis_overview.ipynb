## Interactive Notebook Portfolio Overview

Overview of all notebooks with:
- Technical specifications and requirements
- Dependency analysis and prerequisite validation
- Runtime estimates based on system capabilities
- Output artifacts and business impact metrics

```python
# Display notebook portfolio overview
print("NOTEBOOK PORTFOLIO ANALYSIS")
print("=" * 80)

total_min_time = 0
total_max_time = 0
total_memory_gb = 0

for i, (notebook_id, notebook) in enumerate(nav_hub.notebooks.items(), 1):
    print(f"\n{i:02d}. {notebook.title}")
    print("-" * (len(notebook.title) + 4))
    print(f"    File: {notebook.filename}")
    print(f"    Description: {notebook.description}")
    print(f"    Runtime: {notebook.estimated_runtime_str}")
    print(f"    Memory: {notebook.memory_requirement_gb} GB")
    print(f"    CPU Intensive: {'Yes' if notebook.cpu_intensive else 'No'}")
    print(f"    GPU Recommended: {'Yes' if notebook.gpu_recommended else 'No'}")
    
    print(f"    Prerequisites:")
    for prereq in notebook.prerequisites:
        print(f"      • {prereq}")
    
    print(f"    Key Outputs:")
    for output in notebook.key_outputs:
        print(f"      • {output}")
    
    print(f"    Dependencies: {', '.join(notebook.dependencies)}")
    
    if notebook.output_artifacts:
        print(f"    Output Artifacts:")
        for artifact in notebook.output_artifacts:
            print(f"      • {artifact}")
    
    # Accumulate totals
    min_time, max_time = notebook.estimated_runtime_minutes
    total_min_time += min_time
    total_max_time += max_time
    total_memory_gb += notebook.memory_requirement_gb

print(f"\n{'=' * 80}")
print(f"PORTFOLIO SUMMARY")
print(f"{'=' * 80}")
print(f"Total Notebooks: {len(nav_hub.notebooks)}")
print(f"Total Estimated Time: {total_min_time}-{total_max_time} minutes ({total_min_time/60:.1f}-{total_max_time/60:.1f} hours)")
print(f"Peak Memory Usage: {max(nb.memory_requirement_gb for nb in nav_hub.notebooks.values()):.1f} GB")
print(f"Unique Dependencies: {len(set(dep for nb in nav_hub.notebooks.values() for dep in nb.dependencies))}")
print(f"GPU-Accelerated Notebooks: {sum(1 for nb in nav_hub.notebooks.values() if nb.gpu_recommended)}")
print(f"{'=' * 80}")

# Log portfolio analysis
portfolio_metrics = {
    "total_notebooks": len(nav_hub.notebooks),
    "total_min_time_minutes": total_min_time,
    "total_max_time_minutes": total_max_time,
    "peak_memory_gb": max(nb.memory_requirement_gb for nb in nav_hub.notebooks.values()),
    "gpu_notebooks": sum(1 for nb in nav_hub.notebooks.values() if nb.gpu_recommended)
}

logger.log_data_operation(
    operation="portfolio_analysis",
    status="completed",
    metrics=portfolio_metrics
)
```

## Runtime Estimation and Resource Monitoring

Runtime prediction system featuring:
- **Dynamic System Analysis**: Real-time hardware specification assessment
- **Machine Learning Predictions**: Historical execution data and performance patterns
- **Bottleneck Analysis**: CPU, memory, I/O, and GPU constraint identification
- **Resource Optimization**: Intelligent recommendations for performance improvement
- **Confidence Intervals**: Statistical uncertainty quantification for estimates
- **Dataset Size Scaling**: Adaptive predictions based on data volume

```python
# Runtime Estimation and Resource Monitoring
from utils.runtime_monitor import create_runtime_estimator, SystemSpecs

print("RUNTIME ESTIMATION AND RESOURCE MONITORING")
print("=" * 70)

# Collect detailed system specifications
system_specs = SystemSpecs.collect_current_specs()
print(f"\nSYSTEM SPECIFICATIONS")
print("-" * 30)
print(f"CPU Cores: {system_specs.cpu_count}")
print(f"CPU Frequency: {system_specs.cpu_freq_mhz:.0f} MHz")
print(f"Memory Total: {system_specs.memory_total_gb:.1f} GB")
print(f"Memory Available: {system_specs.memory_available_gb:.1f} GB")
print(f"Disk I/O Speed: {system_specs.disk_io_speed_mbps:.0f} MB/s")
print(f"GPU Available: {'Yes' if system_specs.gpu_available else 'No'}")
if system_specs.gpu_available:
    print(f"GPU Memory: {system_specs.gpu_memory_gb:.1f} GB")

# Estimate dataset size for more accurate predictions
import os
dataset_path = nav_hub.config.data.data_path / nav_hub.config.data.raw_data_file
dataset_size_mb = None
if dataset_path.exists():
    dataset_size_mb = os.path.getsize(dataset_path) / (1024 * 1024)
    print(f"Dataset Size: {dataset_size_mb:.1f} MB")

print(f"\nNOTEBOOK-SPECIFIC RUNTIME ANALYSIS")
print("=" * 50)

total_estimated_time = 0
total_memory_usage = 0
all_recommendations = set()
confidence_scores = []

for notebook_id, notebook in nav_hub.notebooks.items():
    try:
        # Use runtime estimation with dataset size
        estimate = nav_hub.estimate_runtime(notebook_id, dataset_size_mb)
        
        print(f"\n{notebook.title}")
        print("-" * len(notebook.title))
        print(f"   Estimated Duration: {estimate.estimated_duration_minutes:.1f} minutes")
        print(f"   Confidence Level: {estimate.confidence_level:.1%}")
        print(f"   Memory Usage: {estimate.resource_usage['memory_gb']:.1f} GB")
        print(f"   CPU Utilization: {estimate.resource_usage['cpu_utilization']:.1%}")
        print(f"   Disk I/O: {estimate.resource_usage['disk_io_gb']:.1f} GB")
        
        if estimate.optimization_recommendations:
            print(f"   Recommendations:")
            for rec in estimate.optimization_recommendations:
                print(f"     • {rec}")
                all_recommendations.add(rec)
        
        total_estimated_time += estimate.estimated_duration_minutes
        total_memory_usage = max(total_memory_usage, estimate.resource_usage['memory_gb'])
        
        # Log estimation
        logger.log_data_operation(
            operation=f"runtime_estimation_{notebook_id}",
            status="completed",
            metrics={
                "estimated_duration_minutes": estimate.estimated_duration_minutes,
                "confidence_level": estimate.confidence_level,
                "memory_gb": estimate.resource_usage['memory_gb']
            }
        )
        
    except Exception as e:
        print(f"\nError estimating runtime for {notebook.title}: {e}")
        logger.error(f"Runtime estimation failed for {notebook_id}", error=str(e))

print(f"\n{'=' * 60}")
print(f"TOTAL RESOURCE PLANNING")
print(f"{'=' * 60}")
print(f"Total Estimated Time: {total_estimated_time:.1f} minutes ({total_estimated_time/60:.1f} hours)")
print(f"Peak Memory Usage: {total_memory_usage:.1f} GB")
print(f"System Memory Available: {nav_hub.system_info.memory_gb:.1f} GB")
print(f"Memory Utilization: {(total_memory_usage/nav_hub.system_info.memory_gb)*100:.1f}%")

if all_recommendations:
    print(f"\nOPTIMIZATION RECOMMENDATIONS")
    print("-" * 35)
    for rec in sorted(all_recommendations):
        print(f"  • {rec}")

# Resource planning summary
resource_summary = {
    "total_estimated_minutes": total_estimated_time,
    "peak_memory_gb": total_memory_usage,
    "system_memory_gb": nav_hub.system_info.memory_gb,
    "memory_utilization_percent": (total_memory_usage/nav_hub.system_info.memory_gb)*100,
    "optimization_recommendations_count": len(all_recommendations)
}

logger.log_data_operation(
    operation="resource_planning",
    status="completed",
    metrics=resource_summary
)
```
