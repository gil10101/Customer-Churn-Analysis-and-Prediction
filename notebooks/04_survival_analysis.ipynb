# Customer Survival Analysis: Time-to-Event Modeling

## Executive Summary

This notebook implements survival analysis techniques to model customer retention patterns and predict churn timing. The analysis leverages the lifelines library to perform Kaplan-Meier estimation, Cox proportional hazards modeling, and parametric survival analysis.

## Objectives

1. **Survival Function Estimation**: Kaplan-Meier estimator for non-parametric survival analysis
2. **Cox Proportional Hazards**: Semi-parametric modeling with covariate effects
3. **Parametric Models**: Weibull, Log-Normal, and Exponential distribution fitting
4. **Risk Stratification**: Customer segmentation based on survival probabilities
5. **Business Applications**: Customer lifetime value and intervention timing optimization

## Technical Framework

- **Censoring Handling**: Right-censored data treatment for active customers
- **Proportional Hazards Testing**: Schoenfeld residuals and log-rank tests
- **Model Validation**: Concordance index and integrated Brier score
- **Visualization**: Survival curves, hazard functions, and residual analysis
